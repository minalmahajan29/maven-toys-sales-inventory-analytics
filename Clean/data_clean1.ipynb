```python
import pandas as pd
from sqlalchemy import create_engine, text

# 1. Connect to your MySQL database
engine = create_engine('mysql+pymysql://root:minal2004@localhost:3306/maven_toys')

# 2. Wipe the slate clean: Drop old tables safely with foreign keys disabled
with engine.begin() as conn:
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 0;"))
    conn.execute(text("DROP TABLE IF EXISTS sales;"))
    conn.execute(text("DROP TABLE IF EXISTS inventory;"))
    conn.execute(text("DROP TABLE IF EXISTS products;"))
    conn.execute(text("DROP TABLE IF EXISTS stores;"))
    conn.execute(text("DROP TABLE IF EXISTS calendar;"))
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 1;"))

print("Old tables dropped successfully!")

# 3. Process & Import Products
df_prod = pd.read_csv('products.csv')
for col in ['Product_Cost', 'Product_Price']:
    df_prod[col] = df_prod[col].astype(str).str.replace('$', '', regex=False).astype(float)
df_prod.to_sql('products', con=engine, if_exists='replace', index=False)

# 4. Process & Import Stores
df_stores = pd.read_csv('stores.csv')
df_stores['Store_Open_Date'] = pd.to_datetime(df_stores['Store_Open_Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_stores.to_sql('stores', con=engine, if_exists='replace', index=False)

# 5. Process & Import Calendar
df_cal = pd.read_csv('calendar.csv')
df_cal['Date'] = pd.to_datetime(df_cal['Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_cal.to_sql('calendar', con=engine, if_exists='replace', index=False)

# 6. Process & Import Inventory
df_inv = pd.read_csv('inventory.csv')
df_inv.to_sql('inventory', con=engine, if_exists='replace', index=False)

# 7. Process & Import Sales
df_sales = pd.read_csv('sales.csv')
df_sales['Date'] = pd.to_datetime(df_sales['Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_sales.to_sql('sales', con=engine, if_exists='replace', index=False)

print("🚀 Boom! All tables wiped, cleaned, and imported successfully via Python!")
```

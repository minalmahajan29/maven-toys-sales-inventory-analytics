```python
import pandas as pd

# 1. Clean Stores (fix open date format)
df_stores = pd.read_csv('stores.csv')
df_stores['Store_Open_Date'] = pd.to_datetime(df_stores['Store_Open_Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_stores.to_csv('stores_cleaned.csv', index=False)

# 2. Clean Products
df_prod = pd.read_csv('products.csv')
df_prod.to_csv('products_cleaned.csv', index=False)

# 3. Clean Calendar
df_cal = pd.read_csv('calendar.csv')
df_cal['Date'] = pd.to_datetime(df_cal['Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_cal.to_csv('calendar_cleaned.csv', index=False)

# 4. Clean Inventory
df_inv = pd.read_csv('inventory.csv')
df_inv.to_csv('inventory_cleaned.csv', index=False)

# 5. Clean Sales
df_sales = pd.read_csv('sales.csv')
df_sales['Date'] = pd.to_datetime(df_sales['Date'], format='mixed').dt.strftime('%Y-%m-%d')
df_sales.to_csv('sales_cleaned.csv', index=False)

print("All files successfully cleaned and saved with '_cleaned' suffix!")
```

```python
import pandas as pd

# 1. Load products
df_prod = pd.read_csv('products.csv')

# 2. Clean cost and price columns (remove '$', commas, and convert to float)
for col in ['Product_Cost', 'Product_Price']:
    if df_prod[col].dtype == object: # if it's read as text/string
        df_prod[col] = df_prod[col].astype(str).str.replace('$', '', regex=False)
        df_prod[col] = df_prod[col].str.replace(',', '', regex=False).astype(float)

# 3. Save the cleaned version
df_prod.to_csv('products_cleaned.csv', index=False)
print("Products cleaned successfully!")
```
